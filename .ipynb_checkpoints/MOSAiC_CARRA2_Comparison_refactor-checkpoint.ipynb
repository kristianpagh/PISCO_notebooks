# MOSAiC vs CARRA2 comparison — earthkit refactor

Refactor of `MOSAiC_CARRA2_Comparison_v2.ipynb`:

- **GRIB extraction with `earthkit.data`** replaces the `extract_carra2_mosaic.py`
  eccodes subprocess. earthkit indexes the GRIB without decoding it, `.sel()` filters
  on metadata only, and values are decoded one field at a time — peak memory stays
  around one full field (~65 MB), same as the subprocess trick, but in-process and
  a fraction of the code.
- **Nearest grid point with `earthkit.geo`**: a single vectorised
  `nearest_point_haversine` call handles all timesteps at once, replacing the
  per-hour great-circle loop over the full 2-D grid.
- **Plotting with `earthkit.plots`**: figures, layout, time axes and legends are
  managed by `earthkit.plots.Figure`; hexbin and histograms (not yet in
  earthkit-plots) draw onto the earthkit-managed matplotlib axes (`subplot.ax`).
- Consolidated the repeated merge/plot blocks into functions, fixed the
  `base_dir` missing-slash path bug, figures now save to a local `Figures/` folder.

The saved CSV formats are unchanged, so outputs stay compatible with the v2 notebook.

```python
from pathlib import Path

import numpy as np
import pandas as pd
import xarray as xr
import earthkit.data as ekd
from earthkit.geo.distance import nearest_point_haversine
from earthkit.plots import Figure
```

```python
BASE = Path.home() / "Git/PISCO_notebooks"
DATA_DIR = BASE / "Data"
OBS_FILE = DATA_DIR / "mosaic-examples/MOSAiC_obs_combined_1h_v2.nc"

GRIB_DIR = DATA_DIR / "CARRA2"                    # raw monthly CARRA2 GRIBs
SUBSET_DIR = GRIB_DIR / "MOSAiC_Subsets/Monthly"  # cropped monthly netCDFs
FIG_DIR = BASE / "Figures"
FIG_DIR.mkdir(exist_ok=True)

K0 = 273.15
VAR_MAP = {"2t": "t2m", "10si": "wspd10"}         # GRIB shortName -> output name

MONTHS = ['201910', '201911', '201912', '202001', '202002',
          '202003', '202004', '202005', '202006', '202007', '202008']
```

## MOSAiC observations and track

```python
def load_mosaic(time_start, time_end, obs_file=OBS_FILE):
    """Ship track (lat/lon) and hourly 2t / 10si observations for one period."""
    obs = xr.open_dataset(obs_file).sel(time=slice(time_start, time_end))
    track = pd.DataFrame({"lat": obs.lat.to_series(), "lon": obs.lon.to_series()})
    obs_t2m = obs.temp_2m.to_series()               # deg C
    obs_wspd = obs.wspd_vec_mean_10m.to_series()    # m s-1
    for name, s in (("t2m", obs_t2m), ("wspd10", obs_wspd)):
        print(f"{name}: {s.notna().sum()} of {len(s)} hourly obs, "
              f"{s.index[0]} .. {s.index[-1]}")
    return track, obs_t2m, obs_wspd


def track_bbox(track, buffer_km=50):
    """Lat/lon crop bounds around the track, buffered by buffer_km.
    The longitude buffer uses the lowest track latitude (conservative)."""
    lat_buf = buffer_km / 111.32
    lon_buf = buffer_km / (111.32 * np.cos(np.deg2rad(track.lat.min())))
    bbox = (track.lat.min() - lat_buf, track.lat.max() + lat_buf,
            track.lon.min() - lon_buf, track.lon.max() + lon_buf)
    print(f"crop bounds: lat {bbox[0]:.3f} .. {bbox[1]:.3f}, "
          f"lon {bbox[2]:.3f} .. {bbox[3]:.3f}")
    return bbox


def month_bounds(month):
    """'201912' -> ('2019-12-01', '2019-12-31')"""
    t0 = pd.to_datetime(month, format='%Y%m')
    return t0.strftime('%Y-%m-%d'), (t0 + pd.offsets.MonthEnd(1)).strftime('%Y-%m-%d')
```

## CARRA2 subset extraction (`earthkit.data`)

Streams the GRIB field by field: metadata filtering (`sel`, time window) never
decodes data values, and only the bbox slab of each decoded field is kept.
Variables passed together must share a native grid (checked — `sif` lives on a
different grid in this dataset, extract it in a separate call).

```python
def extract_carra2_subset(grib_file, var_map, time_start, time_end, bbox,
                          out_path=None):
    """Crop selected CARRA2 GRIB variables to a bbox and return an xr.Dataset.

    var_map: {shortName: output_variable_name}, e.g. {"2t": "t2m"}.
    time_end is inclusive of the whole end day, as in the v2 notebook.
    """
    start = pd.Timestamp(time_start)
    end = pd.Timestamp(time_end) + pd.Timedelta(days=1)
    min_lat, max_lat, min_lon, max_lon = bbox

    fields = (ekd.from_source("file", str(grib_file))
              .sel(shortName=list(var_map))
              .order_by("valid_datetime"))

    crop = grid_shape = lat_c = lon_c = None
    times = {name: [] for name in var_map.values()}
    slabs = {name: [] for name in var_map.values()}

    for f in fields:
        t = pd.Timestamp(f.metadata("valid_datetime"))
        if not (start <= t < end):
            continue
        shape = (f.metadata("Ny"), f.metadata("Nx"))
        if grid_shape is None:
            grid_shape = shape
        elif shape != grid_shape:
            raise ValueError(
                f"grid mismatch: '{f.metadata('shortName')}' is {shape}, expected "
                f"{grid_shape} — these variables don't share a native grid, "
                f"extract them in separate calls")
        if crop is None:
            ll = f.to_latlon()
            lat2d = ll["lat"]
            lon2d = (ll["lon"] + 180.0) % 360.0 - 180.0
            inside = ((lat2d >= min_lat) & (lat2d <= max_lat) &
                      (lon2d >= min_lon) & (lon2d <= max_lon))
            ys, xs = np.nonzero(inside)
            crop = (slice(ys.min(), ys.max() + 1), slice(xs.min(), xs.max() + 1))
            lat_c, lon_c = lat2d[crop], lon2d[crop]
            print(f"grid {grid_shape} -> crop {lat_c.shape}")
        name = var_map[f.metadata("shortName")]
        slabs[name].append(f.to_numpy()[crop])   # decode one field, keep the slab
        times[name].append(t)

    kept = {name: n for name, s in slabs.items() if (n := len(s))}
    if not kept:
        raise ValueError(f"no {list(var_map)} fields in {time_start}..{time_end}")
    print("analyses kept:", kept)

    ds = xr.merge(
        xr.DataArray(np.stack(slabs[name]), dims=("time", "y", "x"),
                     coords={"time": times[name]}, name=name)
        for name in kept)
    ds = ds.assign_coords(latitude=(("y", "x"), lat_c),
                          longitude=(("y", "x"), lon_c))

    if out_path is not None:
        out_path = Path(out_path)
        if out_path.exists():
            print(f"skip write, already exists: {out_path}")
        else:
            out_path.parent.mkdir(parents=True, exist_ok=True)
            ds.to_netcdf(out_path)
            print(f"saved {out_path}")
    return ds
```

```python
# Extract one month (needs the raw GRIB, e.g. on the external drive) !!!
month = "201912"
grib_file = GRIB_DIR / f"CARRA_{month}_Analysis_Wspd_2mTemp_SIF.grib"

if grib_file.exists():
    t0, t1 = month_bounds(month)
    track, obs_t2m, obs_wspd = load_mosaic(t0, t1)
    carra2 = extract_carra2_subset(
        grib_file, VAR_MAP, t0, t1, track_bbox(track),
        out_path=SUBSET_DIR / f"CARRA2_{month}_MOSAiC_t2m_wspd10.nc")
else:
    print(f"not found: {grib_file}\n"
          f"-> mount the drive with the CARRA2 GRIBs or adjust GRIB_DIR")
```

## Nearest CARRA2 grid point along the track (`earthkit.geo`)

One vectorised `nearest_point_haversine` call maps every analysis time's ship
position to its nearest grid point — no per-timestep loop.

```python
def sample_along_track(carra2, var, track, obs, name="CARRA2"):
    """CARRA2 values at the nearest grid point to the ship, per analysis time.

    Returns a DataFrame with columns ["MOSAiC obs", name], NaN-free.
    """
    da = carra2[var]
    tindex = pd.DatetimeIndex(da.time.values)
    if not tindex.is_unique:
        keep = ~tindex.duplicated()
        print(f"{var}: dropping {(~keep).sum()} duplicate timestamps")
        da = da.isel(time=keep)
        tindex = tindex[keep]

    # ship position at each analysis time (within 1 h)
    pos = track.reindex(tindex, method="nearest", tolerance=pd.Timedelta("1h"))
    ok = pos["lat"].notna().to_numpy()

    idx, dist = nearest_point_haversine(
        (pos["lat"].to_numpy()[ok], pos["lon"].to_numpy()[ok]),
        (carra2.latitude.values.ravel(), carra2.longitude.values.ravel()))

    flat = da.values.reshape(len(tindex), -1)      # (time, y*x)
    vals = np.full(len(tindex), np.nan)
    vals[ok] = flat[np.nonzero(ok)[0], idx]

    series = pd.Series(vals, index=tindex, name=name)
    if series.mean() > 150:                        # K -> deg C if needed
        series -= K0
    print(f"{var}: {series.notna().sum()} values at the floe, "
          f"max distance to nearest grid point: {dist.max() / 1000:.2f} km")
    return pd.DataFrame({"MOSAiC obs": obs, name: series}).dropna()
```

## All months → combined 3-hourly series

Each month needs its cropped subset netCDF (from the extraction step above).
Months without a subset file are skipped. The CSVs keep the v2 format
(`Datetime,Date,MOSAiC obs,CARRA2`) and are only written when at least one
month was processed, so existing files are never clobbered by an empty run.

```python
def combine_months(parts, freq="3h"):
    """Concatenate monthly frames and insert NaN rows on a regular time grid."""
    combined = pd.concat(parts).sort_index()
    combined = combined[~combined.index.duplicated()]
    full = combined.asfreq(freq)
    # 'Date' marks rows that carry real data (NaT on inserted timestamps), as in v2
    full.insert(0, "Date", combined.index.to_series().reindex(full.index))
    return full.rename_axis("Datetime")


t2m_parts, wspd_parts = [], []
for month in MONTHS:
    subset_file = SUBSET_DIR / f"CARRA2_{month}_MOSAiC_t2m_wspd10.nc"
    if not subset_file.exists():
        print(f"skip {month}: {subset_file.name} not found")
        continue
    t0, t1 = month_bounds(month)
    track, obs_t2m, obs_wspd = load_mosaic(t0, t1)
    with xr.open_dataset(subset_file) as carra2:
        t2m_parts.append(sample_along_track(carra2, "t2m", track, obs_t2m))
        wspd_parts.append(sample_along_track(carra2, "wspd10", track, obs_wspd))

if t2m_parts:
    combine_months(t2m_parts).to_csv(DATA_DIR / "Temp2m_CARRA2_MOSAiC.csv")
    combine_months(wspd_parts).to_csv(DATA_DIR / "WindSpeed10m_CARRA2_MOSAiC.csv")
    print("CSVs written")
else:
    print("no monthly subsets processed — plotting will use the existing CSVs")
```

## Statistics

```python
def stats(df, a="MOSAiC obs", b="CARRA2"):
    d = df[[a, b]].dropna()
    return {"r": d[a].corr(d[b]),
            "rmsd": np.sqrt(((d[a] - d[b]) ** 2).mean()),
            "abs_diff": (d[a] - d[b]).abs().mean(),
            "bias": (d[b] - d[a]).mean(),
            "n": len(d)}


def stats_label(s):
    return (f"r: {s['r']:.2f}\nRMSD: {s['rmsd']:.2f}\n"
            f"A.Diff: {s['abs_diff']:.2f}\nn: {s['n']}")
```

## Plotting (from the saved CSVs) — `earthkit.plots`

`earthkit.plots.Figure` handles layout, date axes and legends. Plot types it
does not provide yet (hexbin, histogram) are drawn directly on the
earthkit-managed matplotlib axes via `subplot.ax`.

```python
YLABELS = {"wspd": "10m wind speed (m s$^{-1}$)",
           "t2m": "2m temperature ($^{o}$C)"}
AXLIMS = {"wspd": (0, 20), "t2m": (-45, 5)}


def read_pair_csv(fname):
    return pd.read_csv(DATA_DIR / fname, parse_dates=["Datetime"],
                       index_col="Datetime")


frames = {"wspd": read_pair_csv("WindSpeed10m_CARRA2_MOSAiC.csv"),
          "t2m": read_pair_csv("Temp2m_CARRA2_MOSAiC.csv")}
for df in frames.values():
    df["Diff"] = df["MOSAiC obs"] - df["CARRA2"]

for key, df in frames.items():
    print(key, stats(df))
```

```python
def as_time_da(series):
    """Series with DatetimeIndex -> xarray DataArray with a 'time' coord.
    earthkit.plots reads the date axis from the coordinate; a bare pandas
    Series would be plotted against integer positions."""
    return series.rename_axis("time").to_xarray()


def timeseries_panels(frames, columns, colors, labels, ylabel_suffix="",
                      save=None):
    """Two stacked earthkit time-series panels: wind speed top, temperature below."""
    fig = Figure(rows=2, columns=1, size=(12, 8))
    for key in ("wspd", "t2m"):
        sp = fig.add_subplot()
        for col, color, label in zip(columns, colors, labels):
            if col in frames[key]:
                sp.line(as_time_da(frames[key][col]), label=label, color=color,
                        linewidth=1)
        # ax.set_ylabel: earthkit's ylabel() templating chokes on the {-1} in the LaTeX exponent
        sp.ax.set_ylabel(YLABELS[key].replace(")", f"{ylabel_suffix})")
                         if ylabel_suffix else YLABELS[key], fontsize=12)
        sp.ax.grid(linewidth=0.3)
        if key == "wspd":
            sp.legend()
        else:
            sp.xlabel("Date")
    if save:
        fig.save(str(FIG_DIR / save), dpi=300, bbox_inches="tight")
    fig.show()


timeseries_panels(frames, columns=("MOSAiC obs", "CARRA2"),
                  colors=("steelblue", "darkorange"),
                  labels=("MOSAiC (3hr)", "CARRA2 (3hr)"),
                  save="MOSAiC_CARRA2_Timeseries.png")
```

```python
# Difference timeseries (MOSAiC - CARRA2)
timeseries_panels(frames, columns=("Diff",), colors=("indianred",),
                  labels=("MOSAiC - CARRA2",), ylabel_suffix=" difference",
                  save="MOSAiC_CARRA2_Difference_Timeseries.png")
```

```python
def scatter_panels(frames, kind="hexbin", save=None):
    """Obs-vs-CARRA2 comparison in an earthkit Figure. hexbin has no earthkit
    equivalent yet, so it draws on the earthkit-managed axes (subplot.ax)."""
    fig = Figure(rows=1, columns=2, size=(10, 5))
    for key in ("wspd", "t2m"):
        ax = fig.add_subplot().ax
        d = frames[key].dropna(subset=["MOSAiC obs", "CARRA2"])
        if kind == "hexbin":
            ax.hexbin(d["MOSAiC obs"], d["CARRA2"], gridsize=50, cmap="plasma",
                      mincnt=1)
        else:
            ax.scatter(d["MOSAiC obs"], d["CARRA2"], color="steelblue", s=80,
                       edgecolors="white", linewidths=0.5)
        lo, hi = AXLIMS[key]
        ax.axline((lo, lo), (hi, hi), linestyle="-", color="grey", alpha=0.6,
                  linewidth=1)
        ax.set_xlim(lo, hi)
        ax.set_ylim(lo, hi)
        ax.set_title(YLABELS[key], fontsize=12)
        ax.set_xlabel("MOSAiC")
        ax.set_ylabel("CARRA2")
        ax.grid(linewidth=0.2)
        ax.text(0.97, 0.03, stats_label(stats(d)), transform=ax.transAxes,
                fontsize=11, va="bottom", ha="right")
    if save:
        fig.save(str(FIG_DIR / save), dpi=300, bbox_inches="tight")
    fig.show()


scatter_panels(frames, kind="hexbin", save="MOSAiC_CARRA2_Scatter_Hexbin.png")
scatter_panels(frames, kind="scatter", save="MOSAiC_CARRA2_Scatter.png")
```

```python
def hist_panels(frames, columns, colors, labels, ylim, save=None):
    """Histograms drawn on earthkit-managed axes (no hist in earthkit-plots yet)."""
    fig = Figure(rows=1, columns=2, size=(10, 5))
    for key in ("wspd", "t2m"):
        ax = fig.add_subplot().ax
        for col, color, label in zip(columns, colors, labels):
            ax.hist(frames[key][col].dropna(), histtype="step", density=True,
                    bins=25, edgecolor=color, linewidth=2, label=label)
        ax.set_xlabel(YLABELS[key])
        ax.grid(linewidth=0.2)
        ax.legend()
        ax.set_ylim(ylim)
        if key == "wspd":
            ax.set_ylabel("Normalised Count")
    if save:
        fig.save(str(FIG_DIR / save), dpi=300, bbox_inches="tight")
    fig.show()


hist_panels(frames, columns=("MOSAiC obs", "CARRA2"),
            colors=("steelblue", "darkorange"), labels=("MOSAiC", "CARRA2"),
            ylim=(0, 0.18), save="MOSAiC_CARRA2_Hist.png")

hist_panels(frames, columns=("Diff",), colors=("steelblue",),
            labels=("MOSAiC - CARRA2",), ylim=(0, 0.37),
            save="MOSAiC_CARRA2_Difference_Hist.png")
```

## IFS (Destination Earth) along-track comparison

The four copy-pasted merge blocks of v2 collapse into one loop: each IFS CSV
becomes a named series joined onto the hourly MOSAiC observations.

```python
IFS_FILES = {
    "IFS_2t_ctrl":   ("Temp2m_IFS_Ctrl_MOSAiC.csv", "2t", True),
    "IFS_2t_hist":   ("Temp2m_IFS_Hist_MOSAiC.csv", "2t", True),
    "IFS_10si_ctrl": ("WindSpeed10m_IFS_Ctrl_MOSAiC.csv", "10si", False),
    "IFS_10si_hist": ("WindSpeed10m_IFS_Hist_MOSAiC.csv", "10si", False),
}

obs = xr.open_dataset(OBS_FILE)
ifs = pd.DataFrame({
    "lat": obs.lat.to_series(),
    "lon": obs.lon.to_series(),
    "obs_t2m": obs.temp_2m.to_series(),
    "obs_10si": obs.wspd_vec_mean_10m.to_series(),
}).dropna()

for name, (fname, col, kelvin) in IFS_FILES.items():
    df = pd.read_csv(DATA_DIR / fname, index_col=0)
    s = pd.Series(df[col].values, index=pd.to_datetime(df["time"]), name=name)
    s = s[~s.index.duplicated()]
    if kelvin:
        s = s - K0
    ifs = ifs.join(s, how="left")

ifs_hourly = ifs[~ifs.index.duplicated()].asfreq("1h")
ifs_hourly.tail()
```

```python
# 1-hourly MOSAiC obs on top of the 3-hourly CARRA2 comparison
fig = Figure(rows=2, columns=1, size=(12, 8))

top = fig.add_subplot()
top.line(as_time_da(ifs_hourly["obs_10si"]), label="MOSAiC (1hr)", color="indianred",
         linewidth=1)
top.line(as_time_da(frames["wspd"]["MOSAiC obs"]), label="MOSAiC (3hr)", color="steelblue",
         linewidth=1)
top.line(as_time_da(frames["wspd"]["CARRA2"]), label="CARRA2 (3hr)", color="darkorange",
         linewidth=1)
top.ax.set_ylabel(YLABELS["wspd"], fontsize=12)
top.ax.grid(linewidth=0.3)
top.legend()

bottom = fig.add_subplot()
bottom.line(as_time_da(frames["t2m"]["MOSAiC obs"]), label="MOSAiC (3hr)",
            color="steelblue", linewidth=1)
bottom.line(as_time_da(frames["t2m"]["CARRA2"]), label="CARRA2 (3hr)", color="darkorange",
            linewidth=1)
bottom.ax.set_ylabel(YLABELS["t2m"], fontsize=12)
bottom.xlabel("Date")
bottom.ax.grid(linewidth=0.3)

fig.show()
```
